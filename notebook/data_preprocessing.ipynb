```python
x=df.drop(columns=["battery_id","RUL"])
y=df["RUL"]
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
x_scaled=scaler.fit_transform(x) 
print(x_scaled)
```

```python
x_scaled = pd.DataFrame(x_scaled, columns=x.columns)# this just makes data readable for better debugging
print(x_scaled)
```

```python
x_scaled_df=x_scaled.copy()
x_scaled_df["battery_id"]=df["battery_id"].values
x_scaled_df["RUL"]=df["RUL"].values
print(x_scaled_df)
```

```python
feature_col=[col for col in x_scaled_df.columns if col not in ["battery_id","RUL"]]
target_col='RUL'
print("feature column",feature_col)
print(" target column",target_col)
    
```

```python
x_scaled_df.to_csv("processed_data.csv", index=False)
```
