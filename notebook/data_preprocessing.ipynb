```python
# One RUL value per cycle per battery
cycle_df = df.groupby(["battery_id", "cycle"])["RUL"].mean().reset_index()
cycle_df = cycle_df.sort_values(["battery_id", "cycle"])

signatures = {}
for battery in sorted(cycle_df["battery_id"].unique()):
    bdf = cycle_df[cycle_df["battery_id"] == battery]
    signature = tuple(zip(bdf["cycle"].round(8), bdf["RUL"].round(8)))
    signatures.setdefault(signature, []).append(battery)

overlapping_groups = [group for group in signatures.values() if len(group) > 1]
if overlapping_groups:
    print("Overlapping RUL trends:")
    for group in overlapping_groups:
        print(f"- {', '.join(group)} have the same RUL trend")
else:
    print("No identical RUL trends found.")

overlap_offsets = {}
for group in overlapping_groups:
    center = (len(group) - 1) / 2
    for idx, battery in enumerate(group):
        overlap_offsets[battery] = (idx - center) * 0.8

plt.figure(figsize=(10, 5))
line_styles = ["-", "--", "-.", ":"]
markers = ["o", "s", "^", "D", "x", "P", "*"]

for idx, battery in enumerate(sorted(cycle_df["battery_id"].unique())):
    bdf = cycle_df[cycle_df["battery_id"] == battery]
    y_values = bdf["RUL"] + overlap_offsets.get(battery, 0.0)
    plt.plot(
        bdf["cycle"],
        y_values,
        label=battery,
        linestyle=line_styles[idx % len(line_styles)],
        marker=markers[idx % len(markers)],
        markevery=max(len(bdf) // 12, 1),
        linewidth=2,
        markersize=5,
    )

plt.xlabel("Cycle")
plt.ylabel("RUL (small visual offsets for identical overlaps)")
plt.title("RUL Trend Across All Batteries")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Define the split and the columns used for preprocessing.
# cycle is needed for ordering, but it is not a model feature and should not be scaled.
train_batteries = ["B5", "B6", "B18"]
test_batteries = ["B7"]

MODEL_FEATURE_COLUMNS = ["chI", "chV", "chT", "disI", "disV", "BCt", "SOH"]
ORDER_COLUMN = "cycle"
ID_COLUMN = "battery_id"
TARGET_COLUMN = "RUL"

processed_columns = [ORDER_COLUMN, *MODEL_FEATURE_COLUMNS, ID_COLUMN, TARGET_COLUMN]
model_df = df.drop(columns=["disT"], errors="ignore")[processed_columns].copy()

print("Train batteries:", train_batteries)
print("Test batteries:", test_batteries)
print("Model features to scale:", MODEL_FEATURE_COLUMNS)
print("Kept unscaled:", [ORDER_COLUMN, ID_COLUMN, TARGET_COLUMN])
```

```python
# Split by battery before fitting the scaler.
# the test battery must not influence preprocessing parameters.
train_raw = model_df[model_df[ID_COLUMN].isin(train_batteries)].copy()
test_raw = model_df[model_df[ID_COLUMN].isin(test_batteries)].copy()

overlap = set(train_raw[ID_COLUMN].unique()) & set(test_raw[ID_COLUMN].unique())
if overlap:
    raise ValueError(f"Train/test battery overlap found: {sorted(overlap)}")

print("Train shape before scaling:", train_raw.shape)
print("Test shape before scaling:", test_raw.shape)
print("Train batteries found:", sorted(train_raw[ID_COLUMN].unique()))
print("Test batteries found:", sorted(test_raw[ID_COLUMN].unique()))
```

```python
# Fit MinMaxScaler on training features only, then apply it to train/test/full data.
# this prevents test-data leakage and keeps cycle as a real ordering column.
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaler.fit(train_raw[MODEL_FEATURE_COLUMNS])

full_scaled_features = scaler.transform(model_df[MODEL_FEATURE_COLUMNS])
train_scaled_features = scaler.transform(train_raw[MODEL_FEATURE_COLUMNS])
test_scaled_features = scaler.transform(test_raw[MODEL_FEATURE_COLUMNS])

print("Scaler fitted on train feature columns only.")
```

```python
# Rebuild processed dataframes with scaled features and unscaled cycle/RUL.
# cycle should remain 1, 2, 3... for sorting and window creation.
def build_processed_dataframe(base_df, scaled_features):
    scaled_features_df = pd.DataFrame(
        scaled_features,
        columns=MODEL_FEATURE_COLUMNS,
        index=base_df.index,
    )
    processed_df = pd.concat(
        [
            base_df[[ORDER_COLUMN]].reset_index(drop=True),
            scaled_features_df.reset_index(drop=True),
            base_df[[ID_COLUMN, TARGET_COLUMN]].reset_index(drop=True),
        ],
        axis=1,
    )
    return processed_df[processed_columns]

x_scaled_df = build_processed_dataframe(model_df, full_scaled_features)
train_dataset = build_processed_dataframe(train_raw, train_scaled_features)
test_dataset = build_processed_dataframe(test_raw, test_scaled_features)

print("Processed full shape:", x_scaled_df.shape)
print("Processed train shape:", train_dataset.shape)
print("Processed test shape:", test_dataset.shape)
x_scaled_df.head()
```

```python
# Save the full processed dataset to the shared data folder.
# rerunning the notebook should regenerate the same shared preprocessing files.
from pathlib import Path

current_dir = Path.cwd().resolve()
if (current_dir / "data" / "processed").exists():
    project_root = current_dir
else:
    project_root = current_dir.parent

processed_dir = project_root / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

x_scaled_df.to_csv(processed_dir / "processed_data.csv", index=False)
print("Saved:", processed_dir / "processed_data.csv")
```
